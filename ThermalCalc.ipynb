# Thermal Calculations for heatsinks


Derating Factor:

$ \Delta P = D(T_A - 25^{\circ}C$\)
- $D$ is the Derating Factor.
- $T_A$ is the ambient temperature.

$P_{derated} = P_{D(25^{\circ}C)} - \Delta P$

(can be different from 25, check with datasheets)


HeatSinks:

$\theta_{JA} = \theta_{JC} + \theta_{CS} + \theta_{SA}$ Total thermal resistance - Juntion to Ambient

- $\theta_{JC}$ = Junction to Case
- $\theta_{CS}$ = Case to HeatSink
- $\theta_{SA}$ = HeatSink to Ambient


$P_D = (T_J - T_A)/\theta_{JA} \quad or \quad T_J = P_D\theta_{JA} + T_A$
