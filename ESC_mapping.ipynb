# ESC mapping

Helper script to map throttle values as percentages to duty cycle that the ESC expects.

- The ESC used in this project expects a max 500Hz PWM signal with the following specs:
	- Stop = 1500us $\pm$ 25us
	- Forward (max) = 1900us
	- Reverse (max) = 1100us

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def map_value(x, in_min, in_max, out_min, out_max):
    return (x - in_min) * (out_max - out_min) / (in_max - in_min) + out_min


def map_percentTOperiod(percent):
    return map_value(percent, -100, 100, 1100, 1900)


def map_periodTOpercent(period):
    return map_value(period, 1100, 1900, -100, 100)
```

## Converting Throttle Percent to Period

In the cell below enter the throttle percent you would like to convert:<br>

```python
percent = 3

print(f'throttle percent = {percent:0.1f} %  ->  period = {map_percentTOperiod(percent)} us')
```
